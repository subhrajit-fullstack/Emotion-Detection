```python
# ============================================
# Import Required Libraries
# ============================================

import os
import cv2
import numpy as np
import matplotlib.pyplot as plt

from PIL import Image


# ============================================
# Dataset Path
# ============================================

dataset_path = "Emotion_Dataset"

print("Dataset Path :", dataset_path)


import os

dataset = "/Users/nexusop03/Downloads/Emotion"

classes = []

for item in os.listdir(dataset):
    path = os.path.join(dataset, item)

    # Sirf folders add honge
    if os.path.isdir(path):
        classes.append(item)

print("Classes:", classes)


total = 0

print("Emotion\tImages")
print("-"*25)

for c in classes:

    path = os.path.join(dataset, c)

    count = len([
        img for img in os.listdir(path)
        if img.lower().endswith((".jpg", ".jpeg", ".png"))
    ])

    total += count

    print(f"{c}\t{count}")

print("\nTotal Images:", total)


# ============================================
# Total Emotion Classes
# ============================================

total_classes = len(classes)

print("Total Emotion Classes :", total_classes)


# ============================================
# Display Sample Images
# ============================================

plt.figure(figsize=(15,5))

for index, emotion in enumerate(classes):

    folder_path = os.path.join(dataset, emotion)

    # Sirf image files lo
    images = [img for img in os.listdir(folder_path)
              if img.lower().endswith((".jpg", ".jpeg", ".png"))]

    if len(images) == 0:
        continue

    image_path = os.path.join(folder_path, images[0])

    image = cv2.imread(image_path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    plt.subplot(1, len(classes), index + 1)
    plt.imshow(image)
    plt.title(emotion)
    plt.axis("off")

plt.tight_layout()
plt.show()


# ============================================
# Check Image Information
# ============================================

first_class = classes[0]

folder_path = os.path.join(dataset, first_class)

images = [img for img in os.listdir(folder_path)
          if img.lower().endswith((".jpg", ".jpeg", ".png"))]

image_path = os.path.join(folder_path, images[0])

image = cv2.imread(image_path)

print("Emotion :", first_class)
print("Image Name :", images[0])
print("Image Shape :", image.shape)
print("Image Data Type :", image.dtype)


# ============================================
# Dataset Summary
# ============================================

print("=" * 40)
print("Emotion Detection Dataset Summary")
print("=" * 40)

print("Dataset Path :", dataset)
print("Number of Classes :", total_classes)
print("Classes :", classes)
print("Total Images :", total)

print("=" * 40)
print("Dataset Loaded Successfully.")
```
