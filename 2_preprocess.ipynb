```python
# ============================================
# Import Required Libraries
# ============================================

import os
import cv2
import numpy as np

from sklearn.model_selection import train_test_split
from tensorflow.keras.utils import to_categorical

# ============================================
# Dataset Path
# ============================================

dataset = "/Users/nexusop03/Downloads/Emotion"

classes = []

for item in os.listdir(dataset):

    path = os.path.join(dataset, item)

    if os.path.isdir(path):
        classes.append(item)

classes = sorted(classes)

print(classes)


# ============================================
# Emotion Labels
# ============================================

emotion_label = {}

for index, emotion in enumerate(classes):
    emotion_label[emotion] = index

print(emotion_label)

# ============================================
# Create Empty Lists
# ============================================

images = []
labels = []

# ============================================
# Read Images
# ============================================

IMG_SIZE = 48

for emotion in classes:

    folder_path = os.path.join(dataset, emotion)

    image_files = [img for img in os.listdir(folder_path)
                   if img.lower().endswith((".jpg", ".jpeg", ".png"))]

    for image_name in image_files:

        image_path = os.path.join(folder_path, image_name)

        image = cv2.imread(image_path)

        if image is None:
            continue

        image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

        image = cv2.resize(image, (IMG_SIZE, IMG_SIZE))

        images.append(image)

        labels.append(emotion_label[emotion])

# ============================================
# Convert into NumPy Array
# ============================================

images = np.array(images)

labels = np.array(labels)

print("Images Shape :", images.shape)
print("Labels Shape :", labels.shape)
# ============================================
# Normalize Images
# ============================================

images = images.astype("float32") / 255.0
# ============================================
# Reshape Images
# ============================================

images = images.reshape(-1,48,48,1)

print(images.shape)

# ============================================
# One Hot Encoding
# ============================================

labels = to_categorical(labels)

print(labels.shape)

# ============================================
# Train Test Split
# ============================================

X_train, X_test, y_train, y_test = train_test_split(
    images,
    labels,
    test_size=0.2,
    random_state=42,
    shuffle=True
)

print("Training Images :", X_train.shape)
print("Testing Images :", X_test.shape)

print("Training Labels :", y_train.shape)
print("Testing Labels :", y_test.shape)
# ============================================
# Dataset Ready
# ============================================

print("Preprocessing Completed Successfully.")

# ============================================
# Save Preprocessed Dataset
# ============================================

np.save("X_train.npy", X_train)
np.save("X_test.npy", X_test)

np.save("y_train.npy", y_train)
np.save("y_test.npy", y_test)

print("Dataset Saved Successfully.")
```
